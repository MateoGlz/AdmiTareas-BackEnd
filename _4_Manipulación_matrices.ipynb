```python
import numpy as np

griffith = np.random.randint(1, 11, size=(3, 3))
guts = np.random.randint(1, 11, size=(3, 3))

print("Griffith:")
print(griffith)

print("\nGuts:")
print(guts)

sum_matrix = griffith + guts
print("\nSuma de matrices:")
print(sum_matrix)

subtract_matrix = griffith - guts
print("\nResta de matrices (Griffith - Guts):")
print(subtract_matrix)

multiply_scalar_matrix = griffith * 2
print("\nMultiplicación de Griffith por 2:")
print(multiply_scalar_matrix)

matrix_product = np.dot(griffith, guts)
print("\nMultiplicación matricial (Griffith  * Guts):")
print(matrix_product)
```
