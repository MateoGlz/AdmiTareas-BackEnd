```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('iris')

plt.figure(figsize=(12, 6))

df.groupby('species')[['petal_length', 'petal_width']].mean().plot(kind='bar')

plt.title('Promedio de Largo y Ancho de Pétalos')
plt.ylabel('Centímetros (cm)')
plt.xlabel('Especies')
plt.xticks(rotation=0)
plt.grid(True, alpha=0.3)
plt.legend(['Largo del Pétalo', 'Ancho del Pétalo'])

plt.show()
```

```python
import math
n = len(df['sepal_length'])
bins_auto = math.ceil(math.log2(n)) + 1

print(f"Regla de Sturges: {bins_auto} bins")

plt.figure(figsize=(10, 6))
plt.hist(df['sepal_length'], bins=bins_auto, color='skyblue', edgecolor='black', alpha=0.7)

plt.xlabel('Largo del Sépalo (cm)')
plt.ylabel('Cantidad de Flores')
plt.title('Distribución del Largo de los Sépalos en Flores Iris')

plt.grid(True, alpha=0.3)
plt.show()
```

```python

plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='petal_length', y='petal_width', hue='species', s=60, alpha=0.7)

plt.xlabel('Largo del Pétalo (cm)')
plt.ylabel('Ancho del Pétalo (cm)')
plt.title('Relación entre Largo y Ancho del Pétalo por Especie')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Cargar datos
df = sns.load_dataset('iris')

plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
sns.boxplot(data=df, x='species', y='sepal_length', hue='species', legend=False)
plt.title('Distribución de Largo del Sépalo')
plt.xlabel('Especie')
plt.ylabel('Largo (cm)')

plt.subplot(1, 2, 2)
sns.boxplot(data=df, x='species', y='sepal_width', hue='species', legend=False)
plt.title('Distribución de Ancho del Sépalo')
plt.xlabel('Especie')
plt.ylabel('Ancho (cm)')


plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(12, 6))
plt.plot(df.index, df['petal_length'], color='purple', linewidth=1, alpha=0.7)

plt.xlabel('Número de Observación (Índice)')
plt.ylabel('Longitud del Pétalo (cm)')
plt.title('Tendencia de la Longitud del Pétalo a través de las Observaciones')

plt.grid(True, alpha=0.3)
plt.show()
```
