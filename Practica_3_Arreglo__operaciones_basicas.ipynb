```python
import numpy as np

random_array = np.random.randint(1, 101, 20)
array10 = []
max_val = np.max(random_array)
min_val = np.min(random_array)
values, counts = np.unique(random_array, return_counts=True)
max_count = np.max(counts)
modas = values[counts == max_count]
array10 = [int(num * 10) for num in random_array]

print("Original array:", random_array)
print(f"Max: {max_val}, Min: {min_val}")
print("Multiplied by 10:", array10)
print(f"Moda(s): {modas} (each appears {max_count} times)")
```
