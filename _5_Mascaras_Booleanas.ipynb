```python
import numpy as np
arreglo = np.random.randint(1, 51, 15)
print("Arreglo original:", arreglo)

mascara = arreglo > 25
elementos_mayores_que_25 = arreglo[mascara]
print("Elementos mayores que 25:", elementos_mayores_que_25)

cantidad_mayores_que_25 = np.sum(mascara)
print("Cantidad de elementos mayores que 25:", cantidad_mayores_que_25)
```
