```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

df = sns.load_dataset('tips')

print("\n=== VALORES NULOS ===")
print(df.isnull().sum())
print("\n" + "="*50)

print("=== PRIMERAS 5 FILAS ===")
print(df.head())
print("\n" + "="*50)

print("\n=== INFORMACIÓN DEL DATASET ===")
print(df.info())
print("\n" + "="*50)

print("\n=== ESTADÍSTICAS DESCRIPTIVAS ===")
print(df.describe())
print("\n" + "="*50)

print("\n=== DISTRIBUCIÓN DE VARIABLES CATEGÓRICAS ===")
print("Días de la semana:")
print(df['day'].value_counts())
print("\nMomento del día:")
print(df['time'].value_counts())
print("\nFumadores:")
print(df['smoker'].value_counts())
print("\nGénero:")
print(df['sex'].value_counts())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

df = sns.load_dataset('tips')

correlation_matrix = df[['total_bill', 'tip', 'size']].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix,
            annot=True,
            cmap='inferno',
            center=0,
            square=True,
            linewidths=0.5,
            cbar_kws={'label': 'Coeficiente de Correlación'})

plt.title('Mapa de Calor - Correlación entre Variables Numéricas\nDataset de Propinas', pad=20)
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

print("=== MATRIZ DE CORRELACIÓN ===")
print(correlation_matrix.round(3))
```

```python

df = sns.load_dataset('tips')

plt.figure(figsize=(10, 6))
sns.violinplot(data=df, x='day', y='tip',
               palette='pastel',
               inner='quartile')

plt.title('Distribución de Propinas por Día de la Semana', fontsize=14, pad=20)
plt.xlabel('Día de la Semana', fontsize=12)
plt.ylabel('Propina ($)', fontsize=12)
plt.grid(True, alpha=0.3, axis='y')

plt.xticks(fontsize=11)
plt.yticks(fontsize=11)

plt.tight_layout()
plt.show()

print("=== ESTADÍSTICAS DE PROPINAS POR DÍA ===")
print(df.groupby('day')['tip'].describe().round(2))
```

```python
df = sns.load_dataset('tips')

plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='total_bill', y='tip',
                hue='day',
                palette='Accent',
                s=80,
                alpha=0.7)

plt.title('Relación entre Total de Cuenta y Propina por Día de la Semana', fontsize=14, pad=20)
plt.xlabel('Total de la Cuenta ($)', fontsize=12)
plt.ylabel('Propina ($)', fontsize=12)
plt.grid(True, alpha=0.3)

plt.legend(title='Día de la Semana', title_fontsize=11, fontsize=10)

plt.tight_layout()
plt.show()

correlacion = df['total_bill'].corr(df['tip']).round(3)
print(f"Correlación total entre cuenta y propina: {correlacion}")
```
